```python
import pandas as pd, seaborn as sns
df = sns.load_dataset("tips")
df["tip_pct"] = df.tip / df.total_bill * 100
print(df.shape, df.isna().sum().sum(), "missing values")
df.head()
```

```python
by_day  = df.groupby("day", observed=True)["total_bill"].sum().reset_index()
by_time = df.groupby("time", observed=True)["tip_pct"].mean()
by_size = df.groupby("size")["tip_pct"].mean()

summary = f"""Total revenue: {df.total_bill.sum():.0f}, average tip %: {df.tip_pct.mean():.1f}
Revenue by day: {by_day.set_index('day').total_bill.round(0).to_dict()}
Avg tip % by time: {by_time.round(1).to_dict()}
Avg tip % by party size: {by_size.round(1).to_dict()}"""
print(summary)
```

```python
from google import genai
from google.colab import userdata
import time

client = genai.Client(api_key=userdata.get("GEMINI_KEY"))

prompt = f"""You are a business analyst. Based on this restaurant data summary,
give 3 short insights and 1 recommendation for the owner.\n{summary}"""

insight = None
for name in ["gemini-3.8-flash", "gemini-flash-latest", "gemini-2.5-flash", "gemini-2.5-flash-lite"]:
    for attempt in range(3):
        try:
            insight = client.models.generate_content(model=name, contents=prompt).text
            print("Used model:", name)
            break
        except Exception as e:
            print(f"{name} attempt {attempt+1} failed: {str(e)[:70]}")
            time.sleep(5)
    if insight:
        break


if not insight:
    insight = "AI service busy. Re-run this cell later to generate insights."
print(insight)
```

```python
from plotly.subplots import make_subplots
from IPython.display import Markdown, display

fig = make_subplots(rows=2, cols=2, subplot_titles=(
    "Revenue by Day", "Avg Tip % by Time", "Avg Tip % by Party Size", "Bill vs Tip"))
fig.add_bar(x=by_day.day, y=by_day.total_bill, row=1, col=1)
fig.add_bar(x=by_time.index.astype(str), y=by_time.values, row=1, col=2)
fig.add_bar(x=by_size.index, y=by_size.values, row=2, col=1)
fig.add_scatter(x=df.total_bill, y=df.tip, mode="markers", row=2, col=2)
fig.update_layout(height=650, showlegend=False,
    title=f"Restaurant Dashboard | Revenue: ${df.total_bill.sum():.0f} | Avg Tip: {df.tip_pct.mean():.1f}%")
fig.show()

display(Markdown("## 🤖 AI Insights\n" + insight))
fig.write_html("dashboard.html")
```
